```python
import matplotlib.pyplot as plt
import math

def f(level: int):
    if level == 0:
        return 0
    return round(742 * math.log((level-1)*0.2+1) + 100)


x = list(range(30))
y = [f(i) for i in x]
print(f'{y[1]=}', f'{y[14]=}', f'{y[29]=}', f'{y[29]/y[14]=}')
plt.plot(x[1:], y[1:])
plt.show()
```
